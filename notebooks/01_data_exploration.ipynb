```python
# 1. IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("Libraries imported successfully!")
```

```python
# 2. DEFINE PROJECT PATHS

PROJECT_ROOT = Path.cwd()

# If the notebook is running from the notebooks/ folder,
# move one level up to the project root.
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data"
IBTRACS_DIR = DATA_DIR / "raw" / "ibtracs"

NI_FILE = IBTRACS_DIR / "ibtracs.NI.list.v04r01.csv"
GLOBAL_FILE = IBTRACS_DIR / "ibtracs.since1980.list.v04r01.csv"

print("Project root:", PROJECT_ROOT)
print("IBTrACS folder:", IBTRACS_DIR)
print()
print("NI file exists:", NI_FILE.exists())
print("Global file exists:", GLOBAL_FILE.exists())
```

```python
# 3. CHECK FILE SIZES

def file_size_mb(path):
    return path.stat().st_size / (1024 ** 2)

print(f"NI dataset size: {file_size_mb(NI_FILE):.2f} MB")
print(f"Global dataset size: {file_size_mb(NI_FILE):.2f} MB")
```

```python
# 4. INSPECT RAW CSV HEADER

with open(NI_FILE, 'r', encoding="utf-8", errors="replace") as f:
    for i in range(8):
        print(f.readline().rstrip())
```

```python
# 5. LOAD IBTRACS DATASETS

# Skip row 2 because it contains units, not actual data.
# Row numbering here starts from 0:
# row 0 = column names
# row 1 = units
# row 2 onward = actual cyclone records


ni_df = pd.read_csv(
    NI_FILE, skiprows=[1], low_memory=False
)

global_df = pd.read_csv(
    GLOBAL_FILE, skiprows=[1], low_memory=False
)

print("Datasets loaded successfully!")
print()
print("NI dataset shape: ", ni_df.shape)
print("Gloabal dataset shape: ", global_df.shape)
```

```python
# 6. PREVIEW THE DATA

print("NI Dataset:")
display(ni_df.head())

print("\nGlobal Dataset:")
display(global_df.head())
```

```python
# 7. DATASET STRUCTURE

print("========== NI DATASET ==========")
print("Rows:", ni_df.shape[0])
print("Columns:", ni_df.shape[1])

print("\n========== GLOBAL DATASET ==========")
print("Rows:", global_df.shape[0])
print("Columns:", global_df.shape[1])
```

```python
# 8. COLUMN NAMES

print("Total columns:", len(ni_df.columns))
print()

for i, column in enumerate(ni_df.columns, start=1):
    print(f"{i:3}. {column}")
```

```python
# 9. DATA TYPES

ni_df.info()
```

```python
# 10. BASIC NUMERICAL STATISTICS

ni_df.describe(include="all").T
```

```python
# 11. MISSING VALUE ANALYSIS

missing = ni_df.isna().sum()

missing_summary = pd.DataFrame({
    "missing_count": missing,
    "missing_percentage": (missing / len(ni_df)) * 100
})

missing_summary = missing_summary.sort_values(
    by = "missing_percentage",
    ascending = False
)

display(missing_summary.head(30))
```

```python
# 12. CHECK MISSING-VALUE REPRESENTATIONS


# Look at the most common values in a few important columns
# to see how missing information is represented.

columns_to_check = [
    "WMO_WIND",
    "WMO_PRES",
    "USA_WIND",
    "USA_PRES",
    "USA_SSHS",
    "USA_R34_NE",
    "USA_RMW"
]

for col in columns_to_check:
    print(f"\n--- {col} ---")
    print(ni_df[col].value_counts(dropna=False).head(10))
```

```python
# 13. COUNT WHITESPACE-ONLY VALUES

whitespace_summary = {}

for col in ni_df.columns:
    # Convert to string temporarily and check for blank/space-only values
    whitespace_count = ni_df[col].astype(str).str.strip().eq("").sum()
    whitespace_summary[col] = whitespace_count

whitespace_summary = pd.Series(whitespace_summary).sort_values(
    ascending=False
)

display(
    whitespace_summary.head(30).to_frame("whitespace_only_count")
)
```

```python
# 14. CHECK SPECIAL / PLACEHOLDER VALUES

special_values = ["-999", "-999.0", "999", "NaN", "nan"]

for col in columns_to_check:
    counts = ni_df[col].astype(str).str.strip().isin(special_values).sum()

    if counts > 0:
        print(f"{col}: {counts} special/placeholder values")
```

```python
# 15. DUPLICATE RECORD ANALYSIS

duplicate_count = ni_df.duplicated().sum()

print("Exact duplicate rows: ", duplicate_count)
print("Percentage of duplicate rows: ", 
        round((duplicate_count / len(ni_df)) * 100, 4), "%")
```

```python
# 16. CYCLONE / STORM ID ANALYSIS

unique_cyclones = ni_df["SID"].nunique()

print("Total reocrds:", len(ni_df))
print("Unique cyclone IDs:", unique_cyclones)
print("Average records per cyclone:", 
      round(len(ni_df) / unique_cyclones, 2))
```

```python
# 17. OBSERVATIONS PER CYCLONE

observations_per_cyclone = (
    ni_df.groupby("SID")
    .size()
    .sort_values(ascending=False)
)

print("Cyclones with most observations:")
display(observations_per_cyclone.head(10))

print("\nCyclones with fewest observations:")
display(observations_per_cyclone.tail(10))
```

```python
# 18. DATE / TIME ANALYSIS

dates = pd.to_datetime(
    ni_df["ISO_TIME"],
    errors="coerce"
)
# Why errors="coerce"?
# If a date is invalid, pandas converts it to NaT instead of crashing.
# We're not modifying ni_df yet.

print("Eraliest observation:", dates.min())
print("Latest observation:", dates.max())
print("Invalid dates:", dates.isna().sum())

```

```python
# 19. TIME INTERVAL ANALYSIS

time_differences = (
    ni_df.assign(parsed_time=dates)
    .sort_values(["SID", "parsed_time"])
    .groupby("SID")["parsed_time"]
    .diff()
)

print("Most common time intervals:")
display(time_differences.value_counts().head(10))
```

```python
# 20. GEOGRAPHIC DATA CHECK (Check latitude and longitude)

lat = pd.to_numeric(ni_df["LAT"], errors="coerce")
lon = pd.to_numeric(ni_df["LON"], errors="coerce")

print("LAT")
print("  Minimum:", lat.min())
print("  Maximum:", lat.max())
print("  Invalid/missing:", lat.isna().sum())

print("\nLON")
print("  Minimum:", lon.min())
print("  Maximum:", lon.max())
print("  Invalid/missing:", lon.isna().sum())
```

```python
# 21. INVALID COORDINATE CHECK

invalid_lat = ((lat < -90) | (lat > 90)).sum()
invalid_lon = ((lon < -180) | (lon > 180)).sum()

print("Invalid latitude values:", invalid_lat)
print("Invalid longitude values:", invalid_lon)
```

```python
# 22. IMPORTANT METEOROLOGICAL FIELDS

important_columns = [
    "WMO_WIND",
    "WMO_PRES",

    "USA_WIND",
    "USA_PRES",
    "USA_SSHS",
    "USA_RMW",

    "NEWDELHI_WIND",
    "NEWDELHI_PRES",
    "NEWDELHI_GRADE",

    "BOM_WIND",
    "BOM_PRES",

    "TOKYO_WIND",
    "TOKYO_PRES"
]

summary = []

for col in important_columns:
    if col in ni_df.columns:
        values = ni_df[col].astype(str).str.strip()

        # Count values that are not empty
        non_empty = values.ne("").sum()

        summary.append({
            "column": col,
            "non_empty": non_empty,
            "empty": len(ni_df) - non_empty,
            "data_percentage": round(
                non_empty / len(ni_df) * 100, 2
            )
        })

agency_summary = pd.DataFrame(summary)

display(
    agency_summary.sort_values(
        "data_percentage",
        ascending=False
    )
)
```

```python
# 23. YEAR DISTRIBUTION

year_counts = (
    ni_df["SEASON"]
    .value_counts()
    .sort_index()
)

display(year_counts)


# Check how many records fall within the
# TCIR satellite dataset period: 2003–2016.

tcir_period = ni_df[
    ni_df["SEASON"].between(2003, 2016)
]

print(
    "NI records from 2003–2016:",
    len(tcir_period)
)

print(
    "Percentage of NI dataset:",
    round(len(tcir_period) / len(ni_df) * 100, 2),
    "%"
)
```

```python
# 24. UNIQUE CYCLONES IN TCIR PERIOD

tcir_cyclones = tcir_period["SID"].nunique()

print(
    "Unique NI cyclones from 2003–2016:",
    tcir_cyclones
)
```

```python
# 25. COMPARE NI AND GLOBAL CYCLONE IDs

ni_sids = set(ni_df["SID"].unique())
global_sids = set(global_df["SID"].unique())

common_sids = ni_sids.intersection(global_sids)

print("Unique NI cyclone IDs:", len(ni_sids))
print("Unique Global cyclone IDs:", len(global_sids))
print("NI cyclone IDs also present in Global:", len(common_sids))

print(
    "Percentage of NI cyclone IDs found in Global:",
    round(len(common_sids) / len(ni_sids) * 100, 2),
    "%"
)
```

```python
# 26. BASIN DISTRIBUTION

basin_counts = (
    ni_df["BASIN"]
    .astype(str)
    .str.strip()
    .value_counts()
)

display(basin_counts)
```

```python
# 27. SUB-BASIN DISTRIBUTION

subbasin_counts = (
    ni_df["SUBBASIN"]
    .astype(str)
    .str.strip()
    .value_counts()
)

display(subbasin_counts)
```

```python
# 28. NATURE DISTRIBUTION

nature_counts = (
    ni_df["NATURE"]
    .astype(str)
    .str.strip()
    .value_counts()
)

display(nature_counts)
```

```python
# 29. INTENSITY / PRESSURE FIELD COMPLETENESS

intensity_columns = [
    "WMO_WIND",
    "WMO_PRES",
    "USA_WIND",
    "USA_PRES",
    "NEWDELHI_WIND",
    "NEWDELHI_PRES",
    "BOM_WIND",
    "BOM_PRES",
    "TOKYO_WIND",
    "TOKYO_PRES"
]

summary = []

for col in intensity_columns:
    values = ni_df[col].astype(str).str.strip()
    non_empty = values.ne("").sum()

    summary.append({
        "Field": col,
        "Available": non_empty,
        "Missing": len(ni_df) - non_empty,
        "Available_%": round(non_empty / len(ni_df) * 100, 2)
    })

intensity_summary = pd.DataFrame(summary)

display(
    intensity_summary.sort_values(
        "Available_%",
        ascending=False
    )
)
```

```python
# 30. WIND SOURCE OVERLAP

wind_columns = [
    "USA_WIND",
    "NEWDELHI_WIND",
    "WMO_WIND",
    "TOKYO_WIND"
]

# Convert blank/whitespace values to NaN
wind_data = ni_df[wind_columns].copy()

for col in wind_columns:
    wind_data[col] = (
        wind_data[col]
        .astype(str)
        .str.strip()
        .replace("", np.nan)
    )

# Count how many wind sources are available
wind_data["available_sources"] = wind_data.notna().sum(axis=1)

print("Number of wind sources available per observation:")
display(
    wind_data["available_sources"]
    .value_counts()
    .sort_index()
)

print("\nObservations with at least one wind source:",
      (wind_data["available_sources"] >= 1).sum())

print("Observations with 2 or more wind sources:",
      (wind_data["available_sources"] >= 2).sum())

print("Observations with 3 or more wind sources:",
      (wind_data["available_sources"] >= 3).sum())

print("Observations with all 4 wind sources:",
      (wind_data["available_sources"] == 4).sum())
```

```python
# 31. ESSENTIAL TRACK FIELD QUALITY

essential_columns = [
    "SID",
    "ISO_TIME",
    "LAT",
    "LON"
]

quality_summary = []

for col in essential_columns:
    values = ni_df[col].astype(str).str.strip()

    quality_summary.append({
        "Field": col,
        "Missing": values.eq("").sum(),
        "Available": values.ne("").sum(),
        "Available_%": round(
            values.ne("").sum() / len(ni_df) * 100, 2
        )
    })

quality_df = pd.DataFrame(quality_summary)

display(quality_df)
```

```python
# 32. TRACK TIME INTERVALS

parsed_time = pd.to_datetime(
    ni_df["ISO_TIME"],
    errors="coerce"
)

time_differences = (
    ni_df.assign(parsed_time=parsed_time)
    .sort_values(["SID", "parsed_time"])
    .groupby("SID")["parsed_time"]
    .diff()
)

print("Most common time intervals between observations:")

display(
    time_differences
    .value_counts()
    .head(10)
)
```

# 01 Data Exploration — Summary

This notebook contains the initial exploration of the IBTrACS North Indian Ocean (NI) cyclone track dataset before data cleaning.

## Dataset
- 62,848 observations
- 174 columns
- 1,858 unique cyclones
- Data range: 1842–2025

## Data Quality
- No exact duplicate rows were found.
- `SID`, `ISO_TIME`, `LAT`, and `LON` are 100% available.
- Latitude and longitude values are within valid geographic ranges.
- No invalid dates were found.
- Most observations have a 3-hour time interval.

## Missing Data
- Many agency-specific fields contain blank values.
- Wind and pressure fields have significant missing data.
- `USA_WIND` has the highest availability among the checked wind fields at 26.03%.
- 19,608 observations (31.2%) have at least one available wind source.
- Missing intensity values will not be blindly filled or used as the only reason to remove track observations.

## TCIR Period
- TCIR covers 2003–2016.
- 5,103 NI observations belong to this period.
- These observations represent 132 unique cyclones.
- This period will be important when matching IBTrACS tracks with the TCIR satellite data.

## Conclusion
The core cyclone tracking fields (`SID`, `ISO_TIME`, `LAT`, and `LON`) are complete and reliable for further processing. However, intensity and agency-specific fields contain substantial missing data.

The raw IBTrACS files will remain unchanged. The next stage will clean and standardize the required track fields and save the cleaned data separately in `data/interim/cleaned_tracks/`.
